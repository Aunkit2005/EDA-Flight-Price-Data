```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Data_Train.csv')
df.head()
```

# Data Cleaning

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
#column wise analysis
df.columns
```

```python
df['Airline'].unique()
```

```python
df['Airline'].value_counts()
```

```python
df['Date_of_Journey'].value_counts()
```

```python
df['Date_of_Journey'].str.split('/')
```

```python
df['Days'] = df['Date_of_Journey'].str.split('/').str[0].astype(int)
```

```python
df['Month'] =  df['Date_of_Journey'].str.split('/').str[1].astype(int)
```

```python
df['Years'] =  df['Date_of_Journey'].str.split('/').str[2].astype(int)
```

```python
df.head()
```

```python
del df['Date_of_Journey']
```

```python
df['Source'].value_counts()
```

```python
df['Destination'].value_counts()
```

```python
del df['Route']
```

```python
df.head()
```

```python
df['Dep_Hours'] = df['Dep_Time'].str.split(':').str[0].astype(int)
df['Dep_Minutes'] = df['Dep_Time'].str.split(':').str[1].astype(int)
```

```python
df.head()
```

```python
df['Arrival_Hours'] = df['Arrival_Time']
```

```python
df.head()
```

```python
del df['Arrival_Time']
```

```python
df.head()
```

```python
df['Duration'].value_counts()
```

```python
 df['Duration'] = df['Duration'].replace('5m','0h 5m')
```

```python
df['Duration'].value_counts()
```

```python
df['Duration_hours'] = df['Duration'].str.split(' ').str[0].str.split('h').str[0].astype(int)
```

```python
df['Duration_Minutes']= df['Duration'].str.split(' ').str[1].str.split('m').str[0].fillna(0).astype(int)
```

```python
df.head()
```

```python
df['Total_Duration(Minutes)'] = df['Duration_hours']*60 + df['Duration_Minutes']
```

```python
df.head()
```

```python
del df['Duration']
del df['Total_Duration']
```

```python
df.head()
```

```python
del df['Additional_Info']
```

```python
df['Total_Stops'].unique()
```

```python
 df['Total_Stops'] = df['Total_Stops'].replace('0','non-stop')
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops'].value_counts()
```

# Data Visualization

```python
df.head()
```

```python
## Compute mean price for every airline
df.groupby('Airline')['Price'].mean().plot(kind = 'bar', title = 'Average price for Every Airline')
plt.xlabel('Airline Name')
plt.ylabel('Price')
plt.show()
```

```python
## average price vs no of stops
df.groupby('Total_Stops')['Price'].mean().plot(kind = 'bar')
plt.xlabel('Total Stops')
plt.ylabel('Price')
plt.title('average price vs no of stops')
plt.show()
```

```python
## Duration vs Price
sns.scatterplot(y = 'Price', x = 'Total_Duration(Minutes)', data = df)
plt.title('Duration vs Price')
```

```python
## distribution of price 
sns.distplot(df['Price'])
plt.title('Distribution of price')
```

```python
## outliers in data 
sns.boxplot(df['Price'])
```

```python
## Airline market share
df.value_counts('Airline').sort_values(ascending=False).head(5).plot(kind = 'pie', autopct = '%.2f%%', figsize=(6,6),title = 'Top 5 Airline Companies passenger used')
```

```python
df.value_counts('Destination').plot(kind = 'pie', autopct = '%.2f%%', title = 'Pie Chart of Destination')
```

```python
df.groupby('Month')['Price'].sum().plot(kind = 'bar', title = 'Total Ticket Sales by Month')
```

```python
df.value_counts('Source').plot(kind = 'pie', autopct = '%.2f%%', title = 'Pie Chart of Source Airpot')
```

```python
## avg ticket price based on Destination
df.groupby('Destination')['Price'].mean().plot(kind = 'bar', title = 'avg ticket price based on Destination')
```

```python
df.groupby('Airline')['Price'].sum().plot(kind = 'bar',title = 'Airline-wise Total Revenue')
```
